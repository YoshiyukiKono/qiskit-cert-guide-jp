```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

state = Statevector.from_instruction(qc)

print(state)
```

```python
qc.h(0)

state = Statevector.from_instruction(qc)

print(state)
```

```python
qc.s(0)

state = Statevector.from_instruction(qc)

print(state)
```

出力：

[0.707+0.j 0.+0.707j]

つまり：

|1⟩ に i が付く

```python
qc = QuantumCircuit(1)

qc.h(0)
qc.s(0)
qc.h(0)

state = Statevector.from_instruction(qc)

print(state)
```

ここで：

結果が変わる

理由：

位相が干渉に影響する
