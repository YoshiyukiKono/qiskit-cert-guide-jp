```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

sv = Statevector.from_label('+-')

qc = QuantumCircuit(2)
qc.h([0, 1])

sv_x = sv.evolve(qc)
print(sv_x)
display(sv_x.draw("latex"))
```

```python
bsv = BasisStatevector(sv_x)
display(bsv.draw("latex"))
display(bsv.draw_zbasis())
display(bsv.draw_xbasis())
display(bsv.draw_ybasis())
```
