```python
import numpy as np
from qiskit.quantum_info import Statevector
from IPython.display import display

sv = Statevector(np.array([1, 0, 0, 1j]) / np.sqrt(2))
sv.draw("latex")
```

```python
sv = Statevector(np.array([0, 0, 0, 0]))
sv.draw("latex")
```

```python
sv00 = Statevector(np.array([1, 0, 0, 0]))
sv00.draw("latex")
```

```python
sv00.draw("qsphere")
```

```python
from qiskit.visualization import array_to_latex

array_to_latex(
    sv00.data.reshape(-1, 1),
    prefix=r"|\psi\rangle = ",
)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

sv = Statevector.from_label('+-')

qc = QuantumCircuit(2)
qc.h([0, 1])

sv_x = sv.evolve(qc)
print(sv_x)
```

```python
sv_x.draw("latex")
```
