```python
from basis_statevector import BasisStatevector

import numpy as np
from qiskit.quantum_info import Statevector
from IPython.display import display

from qiskit.circuit.library import XGate
```

```python
sv = BasisStatevector.from_label("0")

display(sv.draw_zbasis())
display(sv.draw_xbasis())
display(sv.draw_ybasis())
```

## Pauliゲートの作用を基底ごとに見る

```python
sv_x = sv.evolve(XGate())
bsv_x = BasisStatevector(sv_x)
```

```python
display(bsv_x.draw_zbasis())
display(bsv_x.draw_xbasis())
display(bsv_x.draw_ybasis())
```

XゲートはZ基底ではbit flipだが、X基底では位相の違いとして見える

## 相対位相が基底を変えると振幅差に変わることを見る

```python
plus = BasisStatevector.from_label("+")
minus = BasisStatevector.from_label("-")

display(plus.draw_zbasis())
display(minus.draw_zbasis())
```

どちらも振幅の絶対値は同じですが、X基底では

```python
display(plus.draw_xbasis())
display(minus.draw_xbasis())
```

として

$$
|+\rangle
$$

と

$$
|-\rangle
$$

に完全に分かれます。

> 位相情報が別の基底では測定確率の差として現れる

ことを、確認できました。


## Y基底で虚数位相の意味を見る

```python
import numpy as np

sv = BasisStatevector(
    [1/np.sqrt(2), 1j/np.sqrt(2)]
)
```

```python
display(sv.draw_zbasis())
display(sv.draw_xbasis())
display(sv.draw_ybasis())
```

Y基底では

$$
|+i\rangle
$$

と非常に単純になります。

これは「$i$ は、Y方向の位相を表している」という直感に結びつきます。

## 固有状態と固有値の理解

Pauli-$X$ の固有状態は

$$
|+\rangle,\ |-\rangle
$$

です。

そこで

```python
sv = BasisStatevector.from_label("-")
```

```python

display(sv.draw_zbasis())
display(sv.draw_xbasis())
display(sv.draw_ybasis())
```

```python
sv_x = sv.evolve(XGate())
bsv_x = BasisStatevector(sv_x)
display(bsv_x.draw_zbasis())
display(bsv_x.draw_xbasis())
display(bsv_x.draw_ybasis())
```

$X$ を作用させると、

$$
X|-\rangle=-|-\rangle
$$

になります。

X基底で見ると、

作用前：

$$
|-\rangle
$$

作用後：

$$
-|-\rangle
$$

です。

状態の「方向」は同じで、係数だけ $-1$ になります。

これで

> 固有ベクトルは、演算しても方向が変わらず、スカラー倍されるだけ

という線形代数の定義が量子状態で直接見えます。

同じことを

* $Z$ と $|0\rangle,|1\rangle$
* $X$ と $|+\rangle,|-\rangle$
* $Y$ と $|+i\rangle,|-i\rangle$

で確認できます。
