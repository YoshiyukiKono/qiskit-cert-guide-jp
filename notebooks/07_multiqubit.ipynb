```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(2)

state = Statevector.from_instruction(qc)

print(state)
```

```python
qc = QuantumCircuit(2)

qc.x(0)
state = Statevector.from_instruction(qc)
print(state)
```

```python
qc = QuantumCircuit(2)

qc.h([0,1])
state = Statevector.from_instruction(qc)
print(state)
```
