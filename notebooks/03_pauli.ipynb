```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)
qc.x(0)

state = Statevector.from_instruction(qc)

print(state)
```

```python
qc = QuantumCircuit(1)

qc.h(0)
qc.z(0)

state = Statevector.from_instruction(qc)

print(state)
```

```python
from qiskit.quantum_info import Operator

qc = QuantumCircuit(1)
qc.x(0)
qc.z(0)

print(Operator(qc))
```

```python
from qiskit.quantum_info import Operator

qc = QuantumCircuit(1)
qc.z(0)
qc.x(0)

print(Operator(qc))
```
